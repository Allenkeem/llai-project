### 관찰
- **부산·울산·제주가 상위로 부상.** 인구 대비 변호사가 상대적으로 많고/사건부담이 낮아, 서울에 가려져 있던 강점이 드러난다.
- **경기는 중위권, 인천은 하위권.** 경기는 인구가 커 변호사 1인당 접근성(A1)이 희석되고, 인천은 변호사 수는 양호하나 사건부담(A2)이 높다 — 어느 쪽도 '수도권 프리미엄'이 없다 → '수도권 우위'는 사실상 **서울 단독 현상**.
- **엔트로피 가중치에서 A3(법률구조) 비중이 커진다.** 서울이 빠지며 A1의 분산이 줄어, 권역 변별에서 취약계층 접근성의 역할이 상대적으로 커진다.
